# Titanic Dataset Exploratory Data Analysis (EDA)

This notebook explores the Titanic dataset to understand its structure, detect patterns, and identify key relationships between variables.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("/content/sample_data/Titanic dataset.csv.zip")

df.head()
```

```python
df.info()
```

```python
df.info()
```

The dataset contains passenger demographic information, ticket details, and survival status. Numerical variables include Age and Fare, while categorical variables include Sex and Passenger Class.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Passenger Age Distribution')
plt.show()
```

Most passengers were between 20 and 40 years old. The distribution indicates that young adults constituted the largest age group aboard the Titanic.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Fare'], bins=30, kde=True)
plt.title('Fare Distribution')
plt.show()
```

Most passengers paid relatively low fares, while a small number paid exceptionally high fares, resulting in a right-skewed distribution.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', data=df)
plt.title('Passenger Class Distribution')
plt.show()
```

Third-class passengers represented the largest proportion of travelers, indicating that most passengers belonged to the lower ticket category.

```python
plt.figure(figsize=(6,4))
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('Survival by Gender')
plt.show()
```

Female passengers exhibited a considerably higher survival rate than male passengers.

```python
plt.figure(figsize=(6,4))
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title('Survival by Passenger Class')
plt.show()
```

Passengers traveling in first class were more likely to survive than those in second and third class.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age vs Survival')
plt.show()
```

Survivors generally appeared slightly younger than non-survivors, although age differences were not extremely pronounced.

```python
survival_counts = df['Survived'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    survival_counts,
    labels=['Did Not Survive','Survived'],
    autopct='%1.1f%%'
)
plt.title('Overall Survival Distribution')
plt.show()
```

The chart shows that a larger proportion of passengers did not survive the disaster.
